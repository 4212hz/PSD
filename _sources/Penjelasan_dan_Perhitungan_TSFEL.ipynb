# Penjelasan dan Perhitungan Manual Fitur TSFEL

Notebook ini berisi penjelasan dan perhitungan manual untuk dua fitur ekstraksi time-series dari pustaka TSFEL (Time Series Feature Extraction Library):
1. `sum_abs_diff(signal)`
2. `wavelet_abs_mean(signal, fs, wavelet)`

Data yang digunakan adalah 5 data pertama dari file **NO2_TanahMerah_timeseries_clean.csv**.

---

## 1. Fitur `sum_abs_diff(signal)`

**Definisi:**
Fitur `sum_abs_diff` menghitung jumlah dari nilai absolut selisih antara titik data yang berurutan dalam sebuah sinyal. Ini mengukur seberapa banyak sinyal berubah atau berfluktuasi dari satu waktu ke waktu berikutnya (ukuran volatilitas).

**Rumus:**
$$ \text{Sum Abs Diff} = \sum_{i=1}^{N-1} |x_{i+1} - x_i| $$

**Perhitungan Manual:**
5 data pertama dari dataset `NO2_TanahMerah_timeseries_clean.csv` (kolom NO2):
- Titik 1 ($x_1$) = $0.00001513$ ($1.512637 \times 10^{-5}$)
- Titik 2 ($x_2$) = $0.00001513$ ($1.512637 \times 10^{-5}$)
- Titik 3 ($x_3$) = $0.00001388$ ($1.387604 \times 10^{-5}$)
- Titik 4 ($x_4$) = $0.00001263$ ($1.262571 \times 10^{-5}$)
- Titik 5 ($x_5$) = $0.00000362$ ($3.616429 \times 10^{-6}$)

**Langkah Perhitungan Selisih Berurutan:**
1. Hari 1 ke Hari 2: $|x_2 - x_1| = |0.00001513 - 0.00001513| = 0.00000000$
2. Hari 2 ke Hari 3: $|x_3 - x_2| = |0.00001388 - 0.00001513| = |-0.00000125| = 0.00000125$
3. Hari 3 ke Hari 4: $|x_4 - x_3| = |0.00001263 - 0.00001388| = |-0.00000125| = 0.00000125$
4. Hari 4 ke Hari 5: $|x_5 - x_4| = |0.00000362 - 0.00001263| = |-0.00000901| = 0.00000901$

**Jumlah Selisih Absolut:**
$$ \text{Total} = 0.00000000 + 0.00000125 + 0.00000125 + 0.00000901 = \mathbf{0.00001151 \text{ mol/m}^2} $$
*(Dalam notasi ilmiah: $1.150994 \times 10^{-5} \text{ mol/m}^2$)*

**Penjelasan Sederhana:** Nilai $0.00001151$ menunjukkan total akumulasi pergerakan/fluktuasi naik-turun konsentrasi NO2 selama 5 hari pertama.

## 2. Fitur `wavelet_abs_mean(signal, fs, wavelet)`

**Definisi:**
Fitur `wavelet_abs_mean` menghitung nilai rata-rata absolut dari Transformasi Wavelet Kontinu (Continuous Wavelet Transform / CWT) dari sinyal. Secara default, `tsfel` menggunakan Ricker wavelet (Mexican hat wavelet) dengan lebar (widths) dari 1 hingga 9. Transformasi wavelet membantu menangkap informasi frekuensi dan waktu secara bersamaan.

**Rumus Utama:**
CWT dari sinyal $x(t)$ dengan wavelet $\psi$ pada skala (width) $a$ adalah konvolusi:
$$ CWT(a, b) = \sum_{t} x(t) \cdot \psi\left(\frac{t - b}{a}\right) $$

**Perhitungan Manual (Contoh untuk Skala $a=1$):**
Menggunakan 5 data pertama:
$x = [0.00001513, 0.00001513, 0.00001388, 0.00001263, 0.00000362]$

Wavelet Ricker pada skala $a=1$ menghasilkan 5 bobot filter:
$$ W = [W_1, W_2, W_3, W_4, W_5] = [-0.3521, 0, 0.8673, 0, -0.3521] $$

**Langkah 1: Konvolusi pada Titik Tengah ($x_3 = 0.00001388$):**
Kalikan setiap titik sinyal dengan bobot filternya:
- $x_1 \times W_1 = 0.00001513 \times (-0.3521) = -0.00000533$
- $x_2 \times W_2 = 0.00001513 \times 0 = 0.00000000$
- $x_3 \times W_3 = 0.00001388 \times 0.8673 = +0.00001204$
- $x_4 \times W_4 = 0.00001263 \times 0 = 0.00000000$
- $x_5 \times W_5 = 0.00000362 \times (-0.3521) = -0.00000127$

Jumlahkan seluruh hasil kali:
$$ CWT(1, 3) = (-0.00000533) + 0 + 0.00001204 + 0 + (-0.00000127) = \mathbf{0.00000544} $$

**Langkah 2: Menghitung Nilai Rata-rata Absolut Skala $a=1$:**
Setelah nilai CWT untuk seluruh titik (1 sampai 5) dihitung dan diabsolutkan:
$$ |CWT(1, :)| = [0.00000445, 0.00000445, 0.00000544, 0.00000396, 0.00000396] $$

Nilai rata-ratanya adalah:
$$ \text{Wavelet Abs Mean}(a=1) = \frac{0.00000445 + 0.00000445 + 0.00000544 + 0.00000396 + 0.00000396}{5} = \mathbf{0.00000445 \text{ mol/m}^2} $$
*(Dalam notasi ilmiah: $4.454 \times 10^{-6} \text{ mol/m}^2$)*

**Penjelasan Sederhana:** Proses ini diulangi untuk skala 1 sampai 9 guna menangkap kekuatan sinyal NO2 pada berbagai tingkat variasi frekuensi harian.
