# Penjelasan dan Perhitungan Manual Fitur TSFEL

Notebook ini berisi penjelasan dan perhitungan manual untuk dua fitur ekstraksi time-series dari pustaka TSFEL (Time Series Feature Extraction Library):
1. `sum_abs_diff(signal)`
2. `wavelet_abs_mean(signal, fs, wavelet)`

Data yang digunakan adalah 5 data pertama dari file **NO2_TanahMerah_timeseries_clean.csv**.

---

## 1. Fitur `sum_abs_diff(signal)`

**Definisi:**
Fitur `sum_abs_diff` menghitung jumlah dari nilai absolut selisih antara titik data yang berurutan dalam sebuah sinyal. Ini mengukur seberapa banyak sinyal berubah atau berfluktuasi dari satu waktu ke waktu berikutnya (ukuran volatilitas).

**Rumus:**
$$ \text{Sum Abs Diff} = \sum_{i=1}^{N-1} |x_{i+1} - x_i| $$

**Perhitungan Manual:**
Misalkan 5 data pertama dari dataset (kolom NO2) adalah:
$x = [0.000015, 0.000015, 0.000014, 0.000013, 0.000004]$

Selisih antar data berurutan:
1. $|x_2 - x_1| = |0.000015 - 0.000015| = 0$
2. $|x_3 - x_2| = |0.000014 - 0.000015| = |-0.000001| = 0.000001$
3. $|x_4 - x_3| = |0.000013 - 0.000014| = |-0.000001| = 0.000001$
4. $|x_5 - x_4| = |0.000004 - 0.000013| = |-0.000009| = 0.000009$

Jumlah selisih absolut:
$$ \text{Total} = 0 + 0.000001 + 0.000001 + 0.000009 = 0.000011 $$

## 2. Fitur `wavelet_abs_mean(signal, fs, wavelet)`

**Definisi:**
Fitur `wavelet_abs_mean` menghitung nilai rata-rata absolut dari Transformasi Wavelet Kontinu (Continuous Wavelet Transform / CWT) dari sinyal. Secara default, `tsfel` menggunakan Ricker wavelet (Mexican hat wavelet) dengan lebar (widths) dari 1 hingga 9. Transformasi wavelet membantu menangkap informasi frekuensi dan waktu secara bersamaan.

**Rumus Utama:**
CWT dari sinyal $x(t)$ dengan wavelet $\psi$ pada skala (width) $a$ adalah konvolusi:
$$ CWT(a, b) = \sum_{t} x(t) \cdot \psi\left(\frac{t - b}{a}\right) $$

**Perhitungan Manual (Contoh untuk Scale / Width $a=1$):**
Menggunakan 5 data asli pertama dari file CSV (tanpa pembulatan):
$x = [1.51263696e-05, 1.51263696e-05, 1.38760401e-05, 1.26257106e-05, 3.61642931e-06]$

Wavelet Ricker didefinisikan sebagai:
$$\psi(t) = \frac{2}{\sqrt{3a}\pi^{1/4}} \left(1 - \frac{t^2}{a^2}\right) e^{-\frac{t^2}{2a^2}}$$

Filter wavelet Ricker untuk $a=1$ memiliki 5 titik ($t = [-2, -1, 0, 1, 2]$).
Filter Ricker: $W = [-0.3521, 0, 0.8673, 0, -0.3521]$

Konvolusi sinyal $x$ dengan $W$ (menggunakan *mode='same'* di bagian tengah):
Untuk titik tengah ($x_3 = 1.38760401e-05$):
$$CWT(1, 3) = (x_1 \times W_1) + (x_2 \times W_2) + (x_3 \times W_3) + (x_4 \times W_4) + (x_5 \times W_5)$$

$$CWT(1, 3) = (1.5126e-05 \times -0.3521) + (0) + (1.3876e-05 \times 0.8673) + (0) + (3.6164e-06 \times -0.3521)$$

$$CWT(1, 3) \approx -5.326e-06 + 0 + 1.2035e-05 + 0 - 1.273e-06 \approx 5.436e-06$$

Proses ini diulangi untuk titik lainnya dan untuk lebar (widths) 1 hingga 9 untuk membentuk matriks CWT (9 x 5). Terakhir, seluruh nilai diabsolutkan dan dicari rata-ratanya.
